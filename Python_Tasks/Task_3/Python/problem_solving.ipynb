# Q1. Find the Frequency of Each Word in a Sentence
**Pseudocode:**
1. Start.
2. Take a sentence as input.
3. Convert the sentence to lowercase.
4. Split the sentence into individual words.
5. Create an empty dictionary to store word frequencies.
6. Visit each word in the sentence.
7. If the word already exists in the dictionary, increase its count by one.
8. Otherwise, add the word to the dictionary with a count of one.
9. Return and display the dictionary.
10. Stop.
Expected output:{'python':2,'is':2,'easy':1,'and':1,'powerful':1}

```python
def q1_word_frequency(sentence):
    words = sentence.lower().split()
    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    return frequency
# Expected example
sentence = "python is easy and python is powerful"
print(q1_word_frequency(sentence))
```

```python

# Additional test case
sentence = "hello world hello python"
print(q1_word_frequency(sentence))
```

# Q2. Find the Frequency of Each Character in a Word
**Pseudocode:**
1. Start.
2. Take a word as input.
3. Create an empty dictionary to store character frequencies.
4. Visit each character in the word.
5. If the character already exists in the dictionary, increase its count by one.
6. Otherwise, add the character to the dictionary with a count of one.
7. Return and display the dictionary.
8. Stop.
Expected output:{'b':1,'a':3,'n':2}

```python
def q2_character_frequency(word):
    frequency = {}

    for char in word:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    return frequency
word = "banana"
print(q2_character_frequency(word))

```

# Q3. Find the First Non-Repeating Character in a Word
**Pseudocode:**
1. Start.
2. Take a word as input.
3. Create an empty dictionary to count each character.
4. Visit every character and store its frequency.
5. Visit the characters again in their original order.
6. If a character has a frequency of one, return that character.
7. If no such character exists, return None.
8. Stop.
Expected output:w

```python
def q3_first_non_repeating(word):
    frequency = {}

    for char in word:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    for char in word:
        if frequency[char] == 1:
            return char

    return None
word = "swiss"
print("Q3 Output:")
print(q3_first_non_repeating(word))
```

```python
# Q3 Additional Test Case
word = "aabb"
print("Q3 Additional Test:")
print(q3_first_non_repeating(word))
```

# Q4. Check Whether Two Strings Are Anagrams
**Pseudocode:**
1. Start.
2. Take two strings as input.
3. Check whether their lengths are equal.
4. If their lengths are different, return False.
5. Sort the characters of both strings.
6. Compare the sorted strings.
7. If they are equal, return True; otherwise, return False.
8. Stop.
Expected output:True

```python
def q4_are_anagrams(s1, s2):
    if len(s1) != len(s2):
        return False

    return sorted(s1) == sorted(s2)
s1 = "listen"
s2 = "silent"

print("Q4 Output:")
print(q4_are_anagrams(s1, s2))
```

```python
# Q4 Additional Test Case
s1 = "hello"
s2 = "world"

print("Q4 Additional Test:")
print(q4_are_anagrams(s1, s2))
```

# Q5. Find the First Repeating Character in a String
**Pseudocode:**
1. Start.
2. Take a string as input.
3. Create an empty set to store visited characters.
4. Read each character from left to right.
5. If the character is already in the set, return that character.
6. Otherwise, add the character to the set.
7. If no character repeats, return None.
8. Stop.
Expected output:r

```python
def q5_first_repeating_char(string):
    seen = set()

    for char in string:
        if char in seen:
            return char
        else:
            seen.add(char)

    return None
string = "programming"

print("Q5 Output:")
print(q5_first_repeating_char(string))
```

```python
# Q5 Additional Test Case
string = "abcde"

print("Q5 Additional Test:")
print(q5_first_repeating_char(string))
```

# Q6. Move All Zeros to the End of a List
**Pseudocode:**
1. Start.
2. Take a list of numbers as input.
3. Create two empty lists: one for non-zero numbers and one for zeros.
4. Visit each number in the original list.
5. If the number is zero, add it to the zeros list.
6. Otherwise, add it to the non-zeros list.
7. Combine the non-zeros list followed by the zeros list.
8. Return the combined list.
9. Stop.
Expected output:[3,1,2,4,0,0,0]

```python
def q6_move_zeros(numbers):
    non_zeros = []
    zeros = []

    for num in numbers:
        if num == 0:
            zeros.append(num)
        else:
            non_zeros.append(num)

    return non_zeros + zeros
numbers = [0, 3, 0, 1, 2, 0, 4]

print("Q6 Output:")
print(q6_move_zeros(numbers))
```

```python
# Q6 Additional Test Case
numbers = [1, 0, 5, 0, 3]

print("Q6 Additional Test:")
print(q6_move_zeros(numbers))
```

# Q7. Find the Second Highest Distinct Number in a List
**Pseudocode:**
1. Start.
2. Take a list of numbers as input.
3. Remove duplicate numbers.
4. Check whether at least two distinct numbers exist.
5. If fewer than two distinct numbers exist, return None.
6. Sort the distinct numbers in descending order.
7. Return the number at index one.
8. Stop.
Expected output:30

```python
def q7_second_highest(numbers):
    unique_numbers = list(set(numbers))

    if len(unique_numbers) < 2:
        return None

    unique_numbers.sort(reverse=True)

    return unique_numbers[1]
numbers = [10, 20, 40, 30, 40, 20]
print("Q7 Output:")
print(q7_second_highest(numbers))
```

```python
# Q7 Additional Test Case
numbers = [5, 5, 5]

print("Q7 Additional Test:")
print(q7_second_highest(numbers))
```

# Q8. Remove Duplicates Without Changing the Original Order
**Pseudocode:**
1. Start.
2. Take a list as input.
3. Create an empty list to store unique elements.
4. Visit each element in the original list.
5. Check whether the element is already in the unique list.
6. If it is not present, add it to the unique list.
7. Return the unique list.
8. Stop.
Expected output:[4,2,5,7]

```python
def q8_remove_duplicates(numbers):
    unique_numbers = []

    for num in numbers:
        if num not in unique_numbers:
            unique_numbers.append(num)

    return unique_numbers
numbers = [4, 2, 4, 5, 2, 7, 5]
print("Q8 Output:")
print(q8_remove_duplicates(numbers))
```

```python
# Q8 Additional Test Case
numbers = [1, 1, 2, 3, 2, 4]

print("Q8 Additional Test:")
print(q8_remove_duplicates(numbers))
```

# Q9. Find the Missing Number from 1 to N
**Pseudocode:**
1. Start.
2. Take a list of numbers and the value N as input.
3. Calculate the sum of all numbers from 1 to N using the formula N × (N + 1) ÷ 2.
4. Calculate the sum of the numbers in the given list.
5. Subtract the list sum from the expected sum.
6. Return the difference as the missing number.
7. Stop.
Expected output:3

```python
def q9_missing_number(nums, N):
    expected_sum = N * (N + 1) // 2
    actual_sum = sum(nums)

    return expected_sum - actual_sum
nums = [1, 2, 4, 5, 6]
N = 6
print("Q9 Output:")
print(q9_missing_number(nums, N))
```

```python
# Q9 Additional Test Case
nums = [1, 3, 4, 5]
N = 5

print("Q9 Additional Test:")
print(q9_missing_number(nums, N))
```

# Q10. Find Two Numbers Whose Sum Equals a Target
**Pseudocode:**
1. Start.
2. Take a list of numbers and a target as input.
3. Create an empty list to store previously visited numbers.
4. Visit each number from left to right.
5. Calculate the complement by subtracting the current number from the target.
6. If the complement exists in the visited list, return the complement and the current number as a pair.
7. Otherwise, add the current number to the visited list.
8. If no valid pair is found, return an empty list.
9. Stop.
Expected output:[2,7]

```python
def q10_two_sum(nums, target):
    seen = []

    for num in nums:
        complement = target - num

        if complement in seen:
            return [complement, num]

        seen.append(num)

    return []
nums = [2, 7, 11, 15]
target = 9
print("Q10 Output:")
print(q10_two_sum(nums, target))
```

```python
# Q10 Additional Test Case
nums = [1, 2, 4]
target = 10

print("Q10 Additional Test:")
print(q10_two_sum(nums, target))
```

# Q11. Find the Number Closest to Zero
**Pseudocode:**
1. Start.
2. Take a list of numbers as input.
3. If the list is empty, return None.
4. Assume the first number is the closest to zero.
5. Visit each number in the list.
6. Compare its absolute value with the absolute value of the current closest number.
7. If it is closer to zero, update the closest number.
8. If both numbers are equally close, choose the positive number.
9. Return the closest number.
10. Stop.
Expected output:2

```python
def q11_closest_to_zero(nums):
    if not nums:
        return None

    closest = nums[0]

    for num in nums:
        if abs(num) < abs(closest):
            closest = num
        elif abs(num) == abs(closest) and num > closest:
            closest = num

    return closest
nums = [-8, -3, 2, 5, -2]
print("Q11 Output:")
print(q11_closest_to_zero(nums))
```

```python
# Q11 Additional Test Case
nums = [-7, -1, 4, 6]

print("Q11 Additional Test:")
print(q11_closest_to_zero(nums))
```

# Q12. Find Common Elements Between Two Lists
**Pseudocode:**
1. Start.
2. Take two lists as input.
3. Create an empty list to store common elements.
4. Visit each element in the first list in its original order.
5. Check whether the element exists in the second list.
6. If it exists in the second list and is not already in the result, add it.
7. Return the result list.
8. Stop.
Expected output:[3, 4, 5]

```python
def q12_common_elements(list1, list2):
    common = []

    for num in list1:
        if num in list2 and num not in common:
            common.append(num)

    return common
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
print("Q12 Output:")
print(q12_common_elements(list1, list2))
```

# Q13. Rotate a List to the Right by K Positions
**Pseudocode:**
1. Start.
2. Take a list and the number of positions K as input.
3. If the list is empty, return an empty list.
4. Calculate the effective rotation using K modulo the list length.
5. If the effective rotation is zero, return a copy of the original list.
6. Separate the last K elements from the remaining elements.
7. Place the last K elements before the remaining elements.
8. Return the rotated list.
9. Stop.
Expected output: [4,5,1,2,3]

```python
def q13_rotate_right(nums, K):
    if not nums:
        return []

    K = K % len(nums)

    return nums[-K:] + nums[:-K] if K != 0 else nums.copy()
nums = [1, 2, 3, 4, 5]
K = 2
print("Q13 Output:")
print(q13_rotate_right(nums, K))
```

```python
# Q13 Additional Test Case
nums = [10, 20, 30]
K = 1

print("Q13 Additional Test:")
print(q13_rotate_right(nums, K))
```

# Q14. Find the Maximum Sum of Two Consecutive Elements
**Pseudocode:**
1. Start.
2. Take a list containing at least two numbers as input.
3. Calculate the sum of the first two elements and store it as the maximum sum.
4. Visit each remaining pair of consecutive elements.
5. Calculate the sum of the current pair.
6. If the current sum is greater than the maximum sum, update the maximum sum.
7. Return the maximum sum.
8. Stop.
Expected output: 14

```python
def q14_max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]

    for i in range(1, len(nums) - 1):
        current_sum = nums[i] + nums[i + 1]

        if current_sum > max_sum:
            max_sum = current_sum

    return max_sum
nums = [4, 7, 2, 9, 5]
print("Q14 Output:")
print(q14_max_consecutive_sum(nums))
```

```python
# Q14 Additional Test Case
nums = [10, 20, 5, 8]

print("Q14 Additional Test:")
print(q14_max_consecutive_sum(nums))
```

# Q15. Find All Pairs of Numbers Whose Sum Equals a Target
**Pseudocode:**
1. Start.
2. Take a list of numbers and a target as input.
3. Create an empty list to store pairs and an empty set to store visited numbers.
4. Visit each number in the list.
5. Calculate its complement by subtracting the current number from the target.
6. If the complement exists in the visited set, create a pair containing the current number and its complement.
7. Arrange the two numbers in ascending order and add the pair to the result.
8. Add the current number to the visited set.
9. Sort all pairs by their first element.
10. Return the list of pairs.
11. Stop.
Expected output: [[1,8],[2,7],[4,5]]

```python
def q15_all_pairs(nums, target):
    pairs = []
    seen = set()

    for num in nums:
        complement = target - num

        if complement in seen:
            pair = sorted([num, complement])

            if pair not in pairs:
                pairs.append(pair)

        seen.add(num)

    pairs.sort(key=lambda pair: pair[0])

    return pairs
nums = [2, 4, 3, 5, 7, 8, 1]
target = 9
print("Q15 Output:")
print(q15_all_pairs(nums, target))
```

```python
# Q15 Additional Test Case
nums = [1, 2, 3, 4]
target = 20

print("Q15 Additional Test:")
print(q15_all_pairs(nums, target))
```
