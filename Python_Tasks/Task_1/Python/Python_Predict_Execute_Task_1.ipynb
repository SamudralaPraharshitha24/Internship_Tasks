# Q35
**Prediction:** [[1, 99], [3, 4]]

**Explanation:** `copy.copy()` creates a shallow copy, so the outer list is copied but the inner lists are still shared. Changing `b[0][1]` therefore also changes the inner list in `a`.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
```

# Q36
**Prediction:** [[1, 2], [3, 4]]

**Explanation:** `copy.deepcopy()` creates a completely independent copy, including the inner lists. So changing `b` does not affect `a`.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```
