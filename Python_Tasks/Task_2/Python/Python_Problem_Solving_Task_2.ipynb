# Q8. Find the Most Frequent Character
#Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.
#Example input:banana
#Expected output:a


```python
#Step 1: Observe
Input: "banana"
Expected output: "a"
The character a appears 3 times, n appears 2 times, and b appears once.

#Step 2: Understand
We need to find the character that appears the most times in a string. If multiple characters have the same highest frequency, return the one encountered first. Ignore spaces.

#Step 3: Break
Store the input string.
Create an empty dictionary to count each character.
Loop through the string and ignore spaces.
Count how many times each character appears.
Find the character with the highest count.
Print that character.

#Step 4: Logic
Use a dictionary to store each character and its frequency.
Use if character != " " to ignore spaces.
Use max() with the dictionary's values to find the most frequent character.
Python's max() returns the first encountered item when the maximum value is tied, provided the items are processed in insertion order.
```

```python
#Step 5: Code
text = "banana"
frequency = {}
for character in text:
    if character != " ":
        frequency[character] = frequency.get(character, 0) + 1
most_frequent = max(frequency, key=frequency.get)
print(most_frequent)
```

#Write a Python program to identify all characters that appear more than once in a string. Display each repeated character only once, in the order of its first appearance. Ignore spaces.
#Example input:programming
#Expected output:['r', 'g', 'm']


```python
#Step 1: Observe
Input: "programming"
Expected output: ['r', 'g', 'm']
The characters r, g, and m appear more than once.
Each repeated character should appear only once in the output, in the order of its first appearance.

#Step 2: Understand
We need to identify all characters that occur more than once in a string. Ignore spaces and maintain the order in which each repeated character first appeared.

#Step 3: Break
Store the input string.
Create an empty dictionary to count character frequencies.
Count each character, ignoring spaces.
Create an empty list for repeated characters.
Check the characters in their original order.
Add a character to the list if its frequency is greater than one and it has not already been added.
Print the list.

#Step 4: Logic
Use a dictionary to count each character's frequency.
Use a list to store repeated characters.
Check frequency[character] > 1 to identify repeated characters.
Use character not in repeated to avoid duplicates.
```

```python
#Step 5: Code
text = "programming"

frequency = {}

for character in text:
    if character != " ":
        frequency[character] = frequency.get(character, 0) + 1

repeated = []

for character in text:
    if character != " " and frequency[character] > 1:
        if character not in repeated:
            repeated.append(character)

print(repeated)
```

```python
#Step 5: Code
text = "Python makes coding useful Python"

words = text.lower().split()

frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

longest = ""

for word in words:
    if frequency[word] == 1:
        if len(word) > len(longest):
            longest = word

print(longest)
```
